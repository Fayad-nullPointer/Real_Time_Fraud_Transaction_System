## Cell 18: Baseline preprocessing and model
Full preprocessing, feature engineering, sampling, training a `RandomForestClassifier` baseline, and printing evaluation metrics. Runs a simple pipeline with imputation and scaling.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

# --- load (skip if already loaded) ---
cust = pd.read_csv('full dataset with brief/customer_profiles.csv')
tx = pd.read_csv('full dataset with brief/synthetic_fraud_transactions.csv')
term = pd.read_csv('full dataset with brief/terminal_profiles.csv')

# --- merge like in your notebook ---
df = tx.merge(cust, on='CUSTOMER_ID', how='left')
df = df.merge(term[['TERMINAL_ID','x_terminal_id','y_terminal_id']], on='TERMINAL_ID', how='left')

# --- datetime features ---
df['TX_DATETIME'] = pd.to_datetime(df['TX_DATETIME'])
df['tx_hour'] = df['TX_DATETIME'].dt.hour
df['tx_dow'] = df['TX_DATETIME'].dt.dayofweek
df['tx_seconds'] = (df['TX_DATETIME'].dt.hour*3600 + df['TX_DATETIME'].dt.minute*60 + df['TX_DATETIME'].dt.second)
df['tx_daynum'] = (df['TX_DATETIME'] - df['TX_DATETIME'].min()).dt.days

# --- distance customer <-> terminal (Euclidean) ---
df['x_customer_id'] = df['x_customer_id'].astype(float)
df['y_customer_id'] = df['y_customer_id'].astype(float)
df['cust_term_dist'] = np.sqrt((df['x_customer_id'] - df['x_terminal_id'])**2 + (df['y_customer_id'] - df['y_terminal_id'])**2)

# --- terminal frequency encoding (simple) ---
term_counts = df['TERMINAL_ID'].value_counts()
df['term_count'] = df['TERMINAL_ID'].map(term_counts).fillna(0)

# --- select features (tweak as desired) ---
features = [
    'AMOUNT', 'tx_hour', 'tx_dow', 'tx_seconds', 'tx_daynum',
    'cust_term_dist',
    'mean_amount', 'std_amount', 'mean_nb_tx_per_day', 'nb_terminals',
    'term_count'
]
# keep only available features
features = [f for f in features if f in df.columns]

# --- target ---
target = 'TX_FRAUD'

# --- sampling to speed up baseline (set sample_frac=1.0 to use all) ---
sample_frac = 0.1
df_sample = df.sample(frac=sample_frac, random_state=42) if sample_frac < 1.0 else df

# --- drop rows with missing target or features entirely missing ---
df_sample = df_sample.dropna(subset=[target])

X = df_sample[features]
y = df_sample[target].astype(int)

# --- train/test split (stratified) ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# --- pipeline: impute numeric -> scale -> classifier ---
num_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

clf = Pipeline([
    ('pre', num_pipe),
    ('rf', RandomForestClassifier(n_estimators=100, max_depth=10, n_jobs=-1, random_state=42))
])

clf.fit(X_train, y_train)

# --- predictions + metrics ---
y_pred = clf.predict(X_test)
y_proba = clf.predict_proba(X_test)[:,1]

print(classification_report(y_test, y_pred, digits=4))
print('ROC AUC:', round(roc_auc_score(y_test, y_proba),4))
print('Confusion matrix:\n', confusion_matrix(y_test, y_pred))
```
